```python
import pandas as pd
import numpy as np
from textblob import TextBlob  # install if missing: pip install textblob

# Load unified data
unified_df = pd.read_csv("../outputs/unified_behavior_dataset.csv")

# Compute sentiment for each review (simple)
def get_sentiment(text):
    try:
        return TextBlob(str(text)).sentiment.polarity
    except:
        return 0.0

unified_df['sentiment'] = unified_df['review_text'].apply(get_sentiment)

# Group by user_id
persona_features = unified_df.groupby('user_id').agg({
    'rating': ['mean', 'std', 'count'],
    'review_text': lambda x: x.astype(str).str.len().mean(),
    'sentiment': ['mean', 'std']
}).reset_index()

# Flatten column names
persona_features.columns = [
    'user_id', 'avg_rating', 'rating_variance', 'review_count',
    'avg_review_length', 'avg_sentiment', 'sentiment_variance'
]

# Fill NaN variance (for users with 1 review)
persona_features['rating_variance'] = persona_features['rating_variance'].fillna(0)
persona_features['sentiment_variance'] = persona_features['sentiment_variance'].fillna(0)

print(f"Persona features shape: {persona_features.shape}")
print(persona_features.head())
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Use persona_features from previous step
# Filter users with at least 5 reviews for meaningful clustering
min_reviews = 5
filtered = persona_features[persona_features['review_count'] >= min_reviews].copy()
print(f"Users with >= {min_reviews} reviews: {len(filtered)}")

# Select features for clustering
features = ['avg_rating', 'rating_variance', 'avg_review_length', 'avg_sentiment', 'sentiment_variance']
X = filtered[features].fillna(0)

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# KMeans (use same n_clusters as before, e.g., 5)
n_clusters = 5
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
filtered['cluster'] = kmeans.fit_predict(X_scaled)

# Analyze clusters to assign archetype names
cluster_means = filtered.groupby('cluster')[features].mean()
print("\nCluster means:")
print(cluster_means)

# Assign archetype names based on interpretation
# You can adjust these labels based on the cluster_means output
def assign_archetype(cluster):
    mapping = {
        0: "Harsh Critic",      # low avg_rating, high variance
        1: "Emotional Storyteller",  # high sentiment, long reviews
        2: "Warm Optimist",      # social sensitivity
        3: "Deep Experience Analyst",  # moderate ratings, high review count
        4: "Reactive Reviewer"  # high avg_rating, low variance
    }
    return mapping.get(cluster, "Balanced Reviewer")

filtered['archetype'] = filtered['cluster'].apply(assign_archetype)

# Also add dominant_value based on archetype (same mapping as before)
archetype_to_dominant = {
    "Harsh Critic": "service_quality",
    "Emotional Storyteller": "ambience",
    "Warm Optimist": "social_proof",
    "Deep Experience Analyst": "food_quality",
    "Reactive Reviewer": "time_efficiency"
}
filtered['dominant_value'] = filtered['archetype'].map(archetype_to_dominant)

# Save new persona dataset
filtered.to_csv("../outputs/persona_dataset_v2.csv", index=False)
print("Saved persona_dataset_v2.csv with shape:", filtered.shape)

# Show sample
print(filtered[['user_id', 'archetype', 'dominant_value', 'review_count', 'avg_rating']].head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Select features for clustering
cluster_features = ['avg_rating', 'rating_variance', 'review_count', 
                    'avg_review_length', 'avg_sentiment', 'sentiment_variance']

X = persona_features[cluster_features].fillna(0)

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# KMeans (adjust n_clusters as needed; 5 is a good start)
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
persona_features['cluster'] = kmeans.fit_predict(X_scaled)

# Select features for clustering
features_for_clustering = ['avg_rating', 'rating_variance', 'avg_review_length', 'avg_sentiment', 'sentiment_variance']
X = persona_features[features_for_clustering].fillna(0)

# Map cluster to archetype based on cluster centers (you can later rename)
# For now, assign generic names
# Group by cluster and compute mean feature values
cluster_profiles = persona_features.groupby('cluster')[features_for_clustering].mean()

print("Cluster profiles:")
print(cluster_profiles)

# Based on the profiles, manually assign archetype names
# Example logic – adjust thresholds based on your actual cluster_profiles output
archetype_map = {}

for i in range(5):  # 5 clusters
    row = cluster_profiles.loc[i]
    avg_r = row['avg_rating']
    var_r = row['rating_variance']
    sent = row['avg_sentiment']
    rev_len = row['avg_review_length']
    
    # Heuristic rules on cluster averages (not per-user)
    if avg_r < 2.5:
        archetype_map[i] = "Harsh Critic"
    elif avg_r > 4.2 and sent > 0.3:
        archetype_map[i] = "Emotional Storyteller"
    elif avg_r > 4.0 and rev_len > 200:
        archetype_map[i] = "Deep Experience Analyst"
    elif var_r > 1.5:
        archetype_map[i] = "Reactive Reviewer"   # high variance = inconsistent ratings
    elif avg_r > 4.0 and sent < 0.2:
        archetype_map[i] = "Warm Optimist"
    else:
        archetype_map[i] = "Balanced Reviewer"

# Apply to persona_features
persona_features['archetype'] = persona_features['cluster'].map(archetype_map)

# Map archetype to dominant_value (using your exact archetypes)
archetype_to_dominant = {
    "Harsh Critic": "service_quality",
    "Emotional Storyteller": "ambience",
    "Deep Experience Analyst": "food_quality",
    "Warm Optimist": "social_proof",
    "Reactive Reviewer": "time_efficiency",
    "Balanced Reviewer": "neutral"
}

persona_features['dominant_value'] = persona_features['archetype'].map(archetype_to_dominant)

# Check mapping
print(persona_features[['cluster', 'archetype', 'dominant_value']].drop_duplicates().sort_values('cluster'))

# Save updated persona file (now with all 18,773 users)
persona_features.to_csv("../outputs/persona_dataset_full.csv", index=False)
print("Saved persona_dataset_full.csv with", len(persona_features), "users")
```
